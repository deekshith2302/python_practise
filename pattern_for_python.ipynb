BASIC STAR PATTERNS TO PRINT


```python
n=int(input())
for i in range(n):
    for j in range(n):
        print("* ",end="")
    print()
```

```python
n = int(input())
for i in range(n+1):
    print("* " * i)
```

```python
n=int(input())
for i in range(1,n+1):
    for j in range(i):
        print("* ",end="")
    print()        
```

```python
n=int(input())
for i in range(n,0,-1):
    for j in range(i):
        print("* ",end="")
    print()
```

```python
n=int(input())
for i in range(1,n+1):
    for j in range(n-i,0,-1): 
        print(" ",end="")
    for k in range(i):
        print("*",end=" ")
    print()
         
```

```python
n = int(input())

for i in range(1, n + 1):

    # Print leading spaces
    for j in range(n - i):
        print(" ", end="")

    # Print stars
    for k in range(i):
        print("*", end=" ")

    # Move to the next line
    print()
```

```python
n=int(input())
for i in range(n,0,-1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(i):
        print("* ",end="")
    print()        
```

```python
##without space after *

n=int(input())
for i in range(n,0,-1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(i):
        print("*",end="")
    print()        
```

```python
n=int(input())
for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()        
```

```python
n=int(input())
for i in range(n,0,-1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()        
```

```python
n=int(input())
for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()        
for i in range(n-1,0,-1):
    for j in range(n-i):
        print(" ",end="")
    for j in range(2*i-1):
        print("*",end="")
    print()        
```
